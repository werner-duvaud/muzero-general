# MuZero General

Run MuZero in Google Colab or Jupyter with Python 3.12 or later. See the [README](https://github.com/werner-duvaud/muzero-general#readme) for configuration and checkpoint evaluation.

The setup cell uses the current checkout when run from the repository root; otherwise it clones the `main` branch into `muzero-general/`. Run the cells in order. The last cell trains CartPole with its default budget and then evaluates the selected checkpoint. This can take some time; follow progress in TensorBoard.

To evaluate the packaged checkpoint instead, replace the last command with `!{sys.executable} muzero.py cartpole --evaluate --episodes 5`.

```python
import os
import subprocess
import sys
import sysconfig
from pathlib import Path

if sys.version_info < (3, 12):
    raise RuntimeError("MuZero requires Python 3.12 or later.")

if not Path("muzero.py").is_file():
    checkout = Path("muzero-general")
    if not checkout.exists():
        subprocess.run(
            [
                "git",
                "clone",
                "--branch",
                "main",
                "https://github.com/werner-duvaud/muzero-general.git",
            ],
            check=True,
        )
    os.chdir(checkout)

%pip install -r requirements.txt
```

```python
# Use TensorBoard installed in this kernel's environment.
tensorboard = Path(sysconfig.get_path("scripts")) / (
    "tensorboard.exe" if os.name == "nt" else "tensorboard"
)
os.environ["TENSORBOARD_BINARY"] = str(tensorboard)

%load_ext tensorboard
%tensorboard --logdir ./results
```
